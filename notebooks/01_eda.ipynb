# 01 — Exploratory data analysis (Gate 2)

Runs `src/eda.py` and shows its tables and figures. The numbers in `reports/eda_summary.md` are regenerated by this run.

```
import subprocess, sys
from pathlib import Path
from IPython.display import Markdown, Image, display
ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
subprocess.run([sys.executable, '-m', 'src.eda'], cwd=ROOT, check=True, capture_output=True)
display(Markdown((ROOT / 'reports/eda_summary.md').read_text()))
```

```
for name in ['class_dist.png', 'tokens_per_class.png', 'nominal_duration_per_class.png', 'duration_per_class.png',
             'rms_per_class.png', 'f0_per_class.png', 'dyn_range_hist.png', 'speaker_emotion.png', 'spectrogram_examples.png']:
    p = ROOT / 'reports/figures' / name
    if p.exists():
        display(Markdown(f'### {name}')); display(Image(filename=str(p)))
```
